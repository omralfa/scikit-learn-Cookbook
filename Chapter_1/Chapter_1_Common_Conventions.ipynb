## 1. Design Philosophy and Estimator

scikit-learn menggunakan API yang konsisten untuk berbagai algoritma machine
learning. Estimator merupakan objek yang mempelajari parameter dari data
melalui method `fit()`.

Setelah proses training, estimator dapat digunakan untuk melakukan prediksi
atau evaluasi.

```python
from sklearn.datasets import load_iris
from sklearn.linear_model import LogisticRegression

X, y = load_iris(return_X_y=True)

model = LogisticRegression(max_iter=200)
model.fit(X, y)

print("Model:", model)
print("Training accuracy:", model.score(X, y))
```

## 2. Transformer

Transformer digunakan untuk melakukan transformasi terhadap data.
Method `fit()` digunakan untuk mempelajari parameter transformasi, sedangkan
`transform()` digunakan untuk mengubah data.

`fit_transform()` menggabungkan kedua proses tersebut.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print("Original shape:", X.shape)
print("Scaled shape:", X_scaled.shape)
print("First row before scaling:", X[0])
print("First row after scaling:", X_scaled[0])
```

## 3. Custom Estimator and Transformer

scikit-learn menggunakan pola API yang memungkinkan komponen khusus dibuat
dengan mengikuti interface yang sama. Komponen tersebut dapat memiliki
parameter, `fit()`, dan method lain seperti `transform()` atau `predict()`.

Pendekatan ini membuat komponen custom dapat digunakan bersama workflow
scikit-learn.


```python
from sklearn.base import BaseEstimator, TransformerMixin

class SimpleScaler(BaseEstimator, TransformerMixin):
    def fit(self, X, y=None):
        self.mean_ = X.mean(axis=0)
        return self

    def transform(self, X):
        return X - self.mean_

custom_scaler = SimpleScaler()
X_custom = custom_scaler.fit_transform(X)

print("Original mean:", X.mean(axis=0))
print("Transformed mean:", X_custom.mean(axis=0))
```

## 4. Pipelines and Workflow Automation

Pipeline menggabungkan beberapa tahap machine learning menjadi satu workflow.
Contohnya adalah menggabungkan preprocessing dengan estimator.

Keuntungannya adalah setiap tahap dapat dijalankan secara konsisten dalam
satu objek.

```python
from sklearn.pipeline import Pipeline

pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(max_iter=200))
])

pipeline.fit(X, y)

print("Pipeline accuracy:", pipeline.score(X, y))
```

## 5. Common Attributes and Methods

Beberapa method umum dalam API scikit-learn:

| Method | Fungsi |
|---|---|
| `fit()` | Melatih atau mempelajari parameter |
| `predict()` | Menghasilkan prediksi |
| `transform()` | Mengubah data |
| `fit_transform()` | Fit dan transform sekaligus |
| `score()` | Mengukur performa model |

Konvensi API yang konsisten membuat berbagai estimator dan transformer dapat
digunakan dengan pola workflow yang serupa.

## 6. Hyperparameter Tuning

Hyperparameter merupakan parameter yang ditentukan sebelum proses training.
`GridSearchCV` dapat digunakan untuk mencoba beberapa kombinasi hyperparameter
dengan cross-validation dan memilih konfigurasi terbaik.

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
    "model__C": [0.1, 1, 10]
}

grid = GridSearchCV(
    pipeline,
    param_grid=param_grid,
    cv=5
)

grid.fit(X, y)

print("Best parameters:", grid.best_params_)
print("Best CV score:", grid.best_score_)
```

## 7. Metadata, Tags, and Best Practices

Metadata dan tags digunakan scikit-learn untuk memberikan informasi mengenai
karakteristik estimator dan bagaimana estimator tersebut dapat digunakan.

Best practices utama dalam penggunaan API scikit-learn adalah:

- Gunakan `fit()` untuk proses pembelajaran.
- Gunakan `transform()` untuk preprocessing.
- Gunakan Pipeline untuk menggabungkan preprocessing dan model.
- Gunakan cross-validation ketika melakukan pemilihan hyperparameter.
- Gunakan API yang konsisten agar workflow mudah dipelihara.
