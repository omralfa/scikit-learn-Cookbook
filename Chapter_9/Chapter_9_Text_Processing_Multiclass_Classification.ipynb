## 1. Landasan Teori

Data teks merupakan data tidak terstruktur yang belum dapat langsung diproses sebagian besar algoritma machine learning. Text vectorization mengubah dokumen menjadi representasi numerik.

- **Bag of Words:** merepresentasikan dokumen berdasarkan frekuensi kata.
- **TF-IDF:** memberikan bobot berdasarkan frekuensi kata dalam dokumen dan kelangkaannya pada kumpulan dokumen.
- **Multiclass classification:** memprediksi satu kelas dari lebih dari dua kelas.
- **Pipeline:** menggabungkan preprocessing dan estimator agar proses pelatihan konsisten.

Vectorizer harus dipelajari hanya dari data training untuk menghindari data leakage.

```python
import numpy as np
import pandas as pd
import sklearn

print("NumPy:", np.__version__)
print("Pandas:", pd.__version__)
print("scikit-learn:", sklearn.__version__)
```

## 2. CountVectorizer dan TF-IDF

CountVectorizer menghitung kemunculan token. TfidfVectorizer memberi bobot lebih rendah pada kata yang terlalu umum dan dapat membantu menonjolkan kata yang lebih informatif.

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

documents = [
    "machine learning mengolah data",
    "data science menggunakan machine learning",
    "model klasifikasi mempelajari data",
    "teks dapat digunakan untuk klasifikasi",
]

count_vectorizer = CountVectorizer()
count_matrix = count_vectorizer.fit_transform(documents)

count_df = pd.DataFrame(
    count_matrix.toarray(),
    columns=count_vectorizer.get_feature_names_out()
)
display(count_df)

tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(documents)

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf_vectorizer.get_feature_names_out()
)
display(tfidf_df.round(3))
```

**Interpretasi:** CountVectorizer menghasilkan hitungan token, sedangkan TF-IDF menghasilkan bobot numerik. Keduanya menghasilkan matriks fitur yang dapat digunakan oleh model klasifikasi.

## 3. Klasifikasi teks multikelas dengan Pipeline

Contoh berikut menggunakan dataset teks kecil buatan sendiri. Tujuannya mendemonstrasikan alur kerja, bukan menghasilkan estimasi performa yang dapat digeneralisasikan ke data nyata.

```python
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

texts = [
    "tim menang pertandingan sepak bola",
    "pemain mencetak gol di stadion",
    "pelatih menyusun strategi pertandingan",
    "saham perusahaan naik di pasar",
    "investor membeli saham perusahaan",
    "harga pasar mengalami kenaikan",
    "komputer menjalankan program python",
    "model machine learning memproses data",
    "perangkat lunak menggunakan algoritma",
    "tim memenangkan laga sepak bola",
    "investasi saham di pasar modal",
    "algoritma komputer memproses informasi",
]
labels = [
    "olahraga", "olahraga", "olahraga",
    "keuangan", "keuangan", "keuangan",
    "teknologi", "teknologi", "teknologi",
    "olahraga", "keuangan", "teknologi",
]

X_train, X_test, y_train, y_test = train_test_split(
    texts, labels, test_size=0.25, random_state=42, stratify=labels
)

text_model = Pipeline([
    ("tfidf", TfidfVectorizer()),
    ("classifier", LogisticRegression(max_iter=1000))
])

text_model.fit(X_train, y_train)
predictions = text_model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, predictions))
print(classification_report(y_test, predictions, zero_division=0))
print("Confusion matrix:")
print(confusion_matrix(
    y_test, predictions, labels=text_model.classes_
))
print("Classes:", text_model.classes_)
```

## 4. Prediksi dokumen baru

```python
new_documents = [
    "pemain mencetak gol dalam pertandingan",
    "investor menganalisis saham perusahaan",
    "algoritma memproses data komputer",
]
for text, prediction in zip(new_documents, text_model.predict(new_documents)):
    print(f"Teks: {text!r} -> Prediksi: {prediction}")
```
